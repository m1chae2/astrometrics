# Remote Image Ingestion: StellarMate & INDI Synchronization

Modern observatories often operate remotely. A low-power single-board computer (such as a **StellarMate** or Raspberry Pi running INDI/Ekos) resides at the telescope pier, capturing FITS frames and saving them to local flash storage.

In this tutorial, you will learn how to:
1. Query remote observatory server connections over TCP/IP using {py:meth}`~wayfindinglib.ObservatoryControl.check_remote_connection`.
2. Scan remote file directories for unassociated FITS exposures via {py:meth}`~wayfindinglib.ObservatoryControl.list_remote_files`.
3. Synchronize newly captured frames into local target directories automatically via {py:meth}`~wayfindinglib.ObservatoryControl.download_remote_frames` as exposures finish.

## 2. Remote Ingestion & Automated Synchronization

The remote ingestion workflow executes asynchronously:
1. **Directory Monitoring**: Polls remote server directories for new `.fits` / `.fit` extensions via {py:meth}`~wayfindinglib.ObservatoryControl.list_remote_files`.
2. **Metadata Inspection**: Reads header keywords (`OBJECT`, `FILTER`, `EXPTIME`) from the remote header before initiating full payload downloads.
3. **Local Target Linking**: Downloads the FITS file payload via {py:meth}`~wayfindinglib.ObservatoryControl.download_remote_frames` and registers it into the local `TargetCatalog` database.
