# 2. Inspecting FITS Metadata

In this tutorial, you will learn how to use the `astrometrics.targets` API to extract FITS header information and check calibration frame statistics.


## 2.1 Extracting FITS Headers

Sometimes you need to inspect the raw FITS header of a specific frame. You can do this using `astrometrics.targets.get_header()`.


```python
from astrometricslib import Astrometrics

astrometrics = Astrometrics()
target = astrometrics.targets.get("M 13")

# Get the first light frame
light_frame = target.frames[0]

# Fetch its header cards
header_cards = astrometrics.targets.get_header(light_frame.path, target)

# Print the first 10 header cards
for card in header_cards[:10]:
    print(f"{card['key']:<8} = {card['value']:<20} / {card['comment']}")
```

## 2.2 Calibration Frame Statistics

Before running the stacking pipeline, you can check if you have the necessary calibration frames (Darks, Flats, Biases) to calibrate your light frames.


```python
stats = astrometrics.targets.get_calibration_frame_statistics(target, target.frames, grouped=True)

for stat in stats:
    print(f"Filter: {stat['filter']}")
    print(f"  Light Frames: {stat['count']} ({stat['exposure']})")
    print(f"  Matched Darks: {stat['darks']}")
    print("-" * 20)
```
