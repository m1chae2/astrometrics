# Observation Execution: Session Queue & Fault Recovery

Robotic observatories execute observation queues autonomously throughout the night. The execution engine ({py:class}`~wayfindinglib.ObservationExecution`) steps through queued targets, manages hardware transitions, records telemetry, and handles unexpected environmental or mechanical faults.

In this tutorial, you will learn how to:
1. Initialize the observation **session queue recorder** ({py:class}`~wayfindinglib.ObservationExecution`).
2. Understand **Meridian Flips**: German Equatorial Mounts (GEM) must rotate $180^\circ$ around the RA axis when a target crosses the local celestial meridian to prevent camera-pier collisions via {py:meth}`~wayfindinglib.ObservationExecution.execute_meridian_flip`.
3. Monitor **divergence telemetry**: Detecting guide star loss, tracking drift, or cloud coverage via {py:meth}`~wayfindinglib.ObservationExecution.record_divergence`.
4. Execute **Safe-State Fault Recovery**: Automatically parking the mount and closing roof/dome enclosures if safety thresholds are breached via {py:meth}`~wayfindinglib.ObservationExecution.recover_fault`.

## 2. Meridian Flips & Safe-State Guardrails

Autonomous execution includes strict safety guardrails:
1. **Pre-Flip Pause**: Pauses exposure sequence 5 minutes before meridian transit.
2. **Pier Side Swap**: Slews mount $180^\circ$ to the opposite side of the pier via {py:meth}`~wayfindinglib.ObservationExecution.execute_meridian_flip`.
3. **Plate-Solve Alignment**: Re-solves field WCS to restore sub-pixel target centering.
4. **Guider Re-lock**: Re-acquires guide star lock and resumes exposure queue.
5. **Emergency Safety Verdict**: If rain/cloud sensors trigger an UNSAFE verdict, the engine aborts active exposures and parks hardware immediately via {py:meth}`~wayfindinglib.ObservationExecution.recover_fault`.
