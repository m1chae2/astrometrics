# Mosaic Sequence Authoring: Multi-Panel Grid Planning

Large astronomical targets—such as the Andromeda Galaxy (**M 31**), spanning over $3^\circ$ across the sky—exceed the field of view (FOV) of most telescope/camera combinations. To capture the full extent of such targets, astronomers shoot **multi-panel mosaics**.

In this tutorial, you will learn how to:
1. Calculate camera sensor **Field of View (FOV)** from focal length ($f$) and detector dimensions ($W \times H$):
   $$\text{FOV} = 2 \arctan\left(\frac{d}{2f}\right)$$
2. Compute multi-panel $R \times C$ rectangular grids (e.g. $2 \times 2$ tiles) using {py:meth}`~wayfindinglib.ObservationPlanning.calculate_panels`.
3. Set panel **overlap percentages** ($10\% - 20\%$) to enable seamless star registration and gradient blending during image stitching.
4. Package multi-tile observation packages using {py:class}`~wayfindinglib.ObservationPlanning`.

---

## 1. Initializing Wayfinder & Generating Mosaic Grid

We calculate a 4-panel ($2 \times 2$) mosaic for target **M 31 Mosaic** with a $15\%$ tile overlap using {py:meth}`~wayfindinglib.ObservationPlanning.calculate_panels`.

```python
from wayfindinglib import Wayfinder

wayfinder = Wayfinder()
target_name = "M 31 Mosaic"
center_ra, center_dec = "00:42:44.3", "+41:16:09"
panels = wayfinder.planning.calculate_panels(
    center_ra=center_ra, center_dec=center_dec, rows=2, cols=2, overlap_percent=15.0
)
print(f"Generated {len(panels)} mosaic tiles for target {target_name}")
```

## 2. Panel Tile Registration & Sequence Authoring

Each tile in `panels` receives its own center coordinates $(\alpha_i, \delta_i)$ offset from the primary target center. The {py:class}`~wayfindinglib.ObservationPlanning` engine packages these tiles into an executable sequence plan specifying filter sequences and exposure counts per panel.
