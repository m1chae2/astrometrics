# Telescope Control: Slewing, Tracking & Guiding Subsystems

Observatory automation requires precise communication with hardware drivers. `wayfindinglib` interfaces with the **INDI (Instrument-Neutral-Distributed-Interface)** protocol over TCP/IP networks to control equatorial mounts, camera sensors, motorized focusers, and filter wheels using {py:class}`~wayfindinglib.ObservatoryControl`.

In this tutorial, you will learn how to:
1. Query INDI hardware device status and safety states via {py:meth}`~wayfindinglib.ObservatoryControl.get_telescope_status`.
2. Issue slew and tracking commands to equatorial mounts at **Sidereal Rate** ($15.041'' / \text{second}$, matching Earth's rotation) via {py:meth}`~wayfindinglib.ObservatoryControl.slew_to_coordinates`.
3. Monitor **Autoguiding** closed-loop pulse corrections (RA/Dec guide errors in arcseconds) via {py:meth}`~wayfindinglib.ObservatoryControl.pulse_guide` to eliminate mount gear periodic error.
4. Adjust stepper motor **focuser positions** ({py:meth}`~wayfindinglib.ObservatoryControl.focus_move`) and **filter wheel** positions ({py:meth}`~wayfindinglib.ObservatoryControl.set_filter`).

---

## 1. Initializing Wayfinder & Checking Mount Status

We query the connected mount hardware for its current tracking state and spatial coordinates via {py:meth}`~wayfindinglib.ObservatoryControl.get_telescope_status`.

```python
from wayfindinglib import Wayfinder

wayfinder = Wayfinder()
status = wayfinder.control.get_telescope_status()
print(
    f"Connected telescope status: {status.get('state', 'IDLE')}, "
    f"RA: {status.get('ra')}, Dec: {status.get('dec')}"
)
```

## 2. Hardware Subsystems & Auto-Guiding Concepts

During long exposure imaging:
- **Mount Tracking**: The Right Ascension motor drives at $15.041'' / \text{sec}$ opposite Earth's spin.
- **Autoguiding**: A guide camera captures rapid sub-exposures ($1-3$ sec), measures guide star centroid drift, and sends millisecond guide pulses ({py:meth}`~wayfindinglib.ObservatoryControl.pulse_guide`) to keep stars locked on exact pixels.
- **Focuser Models**: Temperature-compensated motorized focus moves offset thermal expansion of the telescope optical tube.

We interact with {py:class}`~wayfindinglib.ObservatoryControl` to query guiding corrections.
