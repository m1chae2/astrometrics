# 4. Astrometry & Plate Solving

In this tutorial, you will learn how to run the Astrometry Pipeline. This pipeline detects stars in a master frame, matches them against an online catalog (like Gaia or Tycho), and updates the WCS (World Coordinate System) metadata.


## 4.1 Providing a Master Frame

`run_astrometry()` operates on whatever `target.stacked_image` points at. That path doesn't have to come from `astrometrics.processing.run_stacking()` -- it's a plain field on the `Target` object, so you can assign it directly to any FITS file, including one produced by other software or, as here, a master bundled with the sample data. This is exactly what `run_stacking()`'s own callers do internally once it returns a path.

```python
from astrometricslib import Astrometrics
from astrometricslib.utilities.config_loader import get_configuration

astrometrics = Astrometrics()
target = astrometrics.targets.get("M 13")

sample_data_dir = (
    get_configuration().get_project_root()
    / "documentation" / "notebooks" / "astrometrics" / "sample_data" / "M 13"
)
target.stacked_image = str(sample_data_dir / "M_13_Stacked_ZWO_ASI_533MM_Pro_405mm.fits")
astrometrics.targets.save()
print(f"target.stacked_image = {target.stacked_image}")
```

## 4.2 Running the Astrometry Pipeline

We use `astrometrics.processing.run_astrometry()` to solve the master frame.

```python
print("Starting astrometry pipeline...")
result = astrometrics.processing.run_astrometry(target)
print(f"WCS resolved: {result.get('wcs') is not None}")
```

## 4.3 Inspecting the Output

The pipeline updates the WCS in the FITS header and returns a summary containing quality metrics like the number of stars detected and the pixel scale.

```python
# The pipeline writes the solved WCS directly into the stacked FITS
# file's own header rather than creating a separate frame record, and
# returns a summary of what it found.
print(f"Stars detected: {len(result.get('stellar_objects', []))}")
```
