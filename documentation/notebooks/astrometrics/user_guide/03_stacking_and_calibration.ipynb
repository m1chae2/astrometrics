# 3. Image Stacking & Calibration

In this tutorial, you will learn how to run the Stacking Pipeline. This pipeline calibrates raw light frames with darks and flats, registers (aligns) them, and stacks them into a high signal-to-noise master frame.


## 3.1 Running the Stacking Pipeline

We will use `astrometrics.processing.run_stacking()` to process the Luminance (L) frames for the target M 13.


```python
from astrometricslib import Astrometrics

astrometrics = Astrometrics()
target = astrometrics.targets.get("M 13")

# Run stacking for the Luminance filter
print("Starting stacking pipeline...")
result = astrometrics.processing.run_stacking(target, filter_type="L")
if result:
    # run_stacking returns the output path; it does not update the
    # target itself, so later pipelines (astrometry, photometry) can
    # find the master frame via target.stacked_image.
    target.stacked_image = result
    astrometrics.targets.save()
print(f"Stacked output: {result}")
```

## 3.2 Inspecting the Output

The stacking pipeline returns a `TargetFilesResponse` object (or a dictionary containing it) which contains the paths to the newly generated master frames.


```python
print("Generated Master Frame:")
print(f"  Target: {target.id}")
print(f"  Path: {target.stacked_image}")
```
