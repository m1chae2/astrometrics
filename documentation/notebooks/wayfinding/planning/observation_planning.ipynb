# Observation Planning: Target Visibility & Site Sky Engine

Before slewing a telescope, an astronomer must evaluate whether a target is visible above the horizon, calculate its peak altitude window, and avoid atmospheric extinction.

In this tutorial, you will learn how to:
1. Define observatory **geodetic coordinates** (latitude $\phi$, longitude $\lambda$, elevation $h$).
2. Convert celestial equatorial coordinates $(\alpha, \delta)$ to local **horizontal coordinates**: Altitude ($a$) and Azimuth ($A$):
   $$\sin a = \sin \phi \sin \delta + \cos \phi \cos \delta \cos H$$
   where $H = \text{LST} - \alpha$ is the Local Hour Angle.
3. Evaluate **Airmass** ($X \approx \csc a$) and enforce minimum altitude safety limits ($a \ge 30^\circ$) using {py:meth}`~wayfindinglib.ObservationPlanning.get_visibility`.
4. Identify **Meridian Transit** windows ($H=0$) for optimal imaging.

---

## 1. Initializing Wayfinder & Planning High-Level Interface

The {py:class}`~wayfindinglib.Wayfinder` high-level interface provides high-level APIs ({py:class}`~wayfindinglib.ObservationPlanning`) for site altitude calculations.

```python
from wayfindinglib import Wayfinder

wayfinder = Wayfinder()
target_name = "M 81"
resolved = wayfinder.planning.resolve_target_coordinates(target_name)
visibility = wayfinder.planning.get_visibility([resolved])
print(f"Target {target_name} visibility details: Altitude {visibility[0].get('altitude', 45.0)}°")
```

## 2. Airmass & Atmospheric Extinction Concepts

At lower altitudes ($a < 30^\circ$), starlight passes through a thicker column of Earth's atmosphere ($X > 2.0$), causing:
- Increased atmospheric extinction (photons absorbed by air molecules).
- Degraded seeing (thermal turbulence smearing star FWHM).
- Rayleigh blue light scattering.

Observation plans prioritize targets near their **Meridian Transit**, where airmass reaches its minimum $X \approx 1.0$, evaluated via {py:meth}`~wayfindinglib.ObservationPlanning.get_visibility`.
